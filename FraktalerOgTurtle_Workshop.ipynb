## DEL 2: Forstå Punkter og Koordinater

Før vi går videre, skal vi forstå hvordan vi arbejder med punkter i Python

Et punkt i 2D har to koordinater: `(x, y)`
- x: hvor langt til højre eller venstre
- y: hvor højt op eller ned

Eksempler:
- `(0, 0)` er midten af skærmen
- `(-100, 50)` er 100 pixels til venstre og 50 pixels opad
- `(100, -50)` er 100 pixels til højre og 50 pixels nedad

En trekant defineres af tre punkter, f.eks.:

`trekant = [(0, 100), (-100, -50), (100, -50)]`

Dette ville lave en trekant med:
- Top punkt: (0, 100)
- Venstre hjørne: (-100, -50)
- Højre hjørne: (100, -50)
